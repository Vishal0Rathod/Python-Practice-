```python
df = pd.read_csv('/Users/w.ishal27/Downloads/Airline1 (1).csv')
df.isna().sum()
df['Source'].value_counts()
source_values = df['Source'].value_counts()
source_values
# value_counts() returns a Series indexed by source names, not by integer position
most_frequent_source = df['Source'].mode().iloc[0]
df['Source'] = df['Source'].fillna(most_frequent_source)
```

```python
dest_values = df['Destination'].mode()

df['Destination'] = df['Destination'].fillna(dest_values.index[0])  # Fill NaN values with the most frequent value
df.isna().sum()
```

```python
Mean_price = df['Price'].mean()
df['Price'] = df['Price'].fillna(Mean_price)  # Fill NaN values with the mean value
df.isna().sum()
```
