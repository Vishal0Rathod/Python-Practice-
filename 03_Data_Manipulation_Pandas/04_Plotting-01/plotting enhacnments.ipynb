```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel('/Users/w.ishal27/Downloads/Swarn_Airline.xlsx') 
df.head(2)
```

```python
sns.set_theme(style="darkgrid", palette="Set1")
plt.figure(figsize=(10,5))

ax = sns.barplot(x=df['Airline'], y=df['Price'], errorbar=None, estimator='sum')
ax.bar_label(ax.containers[0])

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="darkgrid", palette="Set1")
plt.figure(figsize=(10,5))

ax = sns.barplot(x=df['Airline'], y=df['Price'], errorbar=None, estimator='sum', hue=df['Source'])
ax.bar_label(ax.containers[0])

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="darkgrid", palette="Set1")
plt.figure(figsize=(10,5))

ax = sns.barplot(x=df['Airline'], y=df['Price'], errorbar=None, estimator='max', hue=df['Source'])
ax.bar_label(ax.containers[0], rotation=90)

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="darkgrid", palette="Set1")
plt.figure(figsize=(10,5))

ax = sns.barplot(x=df['Airline'], y=df['Price'], errorbar=None, estimator='max', legend=False)
ax.bar_label(ax.containers[0], rotation=90)

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="darkgrid", palette="Set1")
plt.figure(figsize=(10,5))
Airline_Sort = df.groupby('Airline')['Price'].sum().sort_values().index

ax = sns.barplot(x=df['Airline'], y=df['Price'], errorbar=None, estimator='sum', legend=False, order=Airline_Sort, alpha=.9, gap=.1)
ax.bar_label(ax.containers[0], rotation=90)

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="darkgrid", palette="Set1")
plt.figure(figsize=(10,5))
Airline_Sort = df.groupby('Airline')['Price'].count().sort_values().index

ax = sns.countplot(x=df['Airline'],legend=False, order=Airline_Sort, alpha=.9, gap=.1)
ax.bar_label(ax.containers[0], rotation=90)

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python

# Group by Airline and calculate minimum price
newDf = df.groupby('Airline')['Price'].min().sort_values().reset_index()

# Set theme and palette
sns.set_theme(style="whitegrid", palette="Set2")

plt.figure(figsize=(12, 6))


ax = sns.barplot(
    data=newDf,
    x='Airline',
    y='Price',
    hue='Airline',
    palette='Set2',
    legend=False,
    order=newDf.sort_values('Price')['Airline'],
    alpha=0.9,
    gap=0.1,
    errorbar=None
)

ax.bar_label(
    ax.containers[0],
    fmt='%.0f',
    padding=4,
    fontsize=10,
    fontweight='bold'
)

plt.title('Minimum Ticket Price by Airline',
          fontsize=16, fontweight='bold', pad=20)

plt.xlabel('Airline', fontsize=12, fontweight='bold')
plt.ylabel('Minimum Price (₹)', fontsize=12, fontweight='bold')


plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(fontsize=10)

ax.grid(axis='y', linestyle='--', alpha=0.4)
ax.grid(axis='x', visible=False)


sns.despine()

plt.figtext(
    0.99, 0.01,
    'Source: Airline Dataset',
    ha='right',
    fontsize=9,
    color='gray'
)

plt.tight_layout()
plt.show()
```
