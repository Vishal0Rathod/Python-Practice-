```python
#Read Data Here
import pandas as pd


df = pd.read_csv('/Users/w.ishal27/Downloads/Airline1 (2).csv')
df 
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 5))

sns.lineplot( data=df, x='Airline', y='Price', marker='o')

plt.title('Price by Airline')
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(15,5))
sns.violinplot(x=df['Price'])
```

```python
sns.scatterplot(df['Price'])
```

```python
df.head()
```

```python
plt.Figure(figsize=(10,4))
ax = sns.countplot(x=df['Airline'])
ax.bar_label(ax.containers[0], fmt='%d', fontsize=8)
plt.xticks(rotation=90)
plt.show()
```

```python
# Count plot
plt.figure(figsize=(8,4))
ax = sns.countplot(data=df, x='Destination')
ax.bar_label(ax.containers[0], fmt='%d', label_type='edge', fontsize=8)
plt.xticks(rotation=30)

font = 15
plt.xlabel('Destination', fontsize=font)
plt.ylabel('Count', fontsize=font)
plt.title("Count Plot", fontsize=font)
plt.show()
```

```python
plt.Figure(figsize=(10,4))
ax = sns.countplot(x=df['Source'])
ax.bar_label(ax.containers[0], fmt='%d', fontsize=8)
plt.xticks(rotation=90)
plt.show()
```

```python
plt.Figure(figsize=(10,4))
ax = sns.countplot(x=df['Route'])
ax.bar_label(ax.containers[0], fmt='%d', fontsize=8)
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize = (35,5))

ax1 = sns.countplot(x = df['Route'])
ax1.bar_label(ax1.containers[0], fmt = '%d', fontsize = 8)

plt.xticks(rotation = 90)
plt.show()
```

```python
plt.Figure(figsize=(10,3))
sns.lineplot(x=df['Date_of_Journey'], y=df['Price'], marker='o')

plt.title("Airline VS Price")
plt.xticks(rotation=90)
plt.show()
```

```python

sns.swarmplot(y=df['Price'])
```

```python
plt.figure(figsize=(15,6))
sns.boxenplot(data=df, x='Airline', y='Price')
plt.show()
```
