```python
#adding elements to a set
mySet = {"apple", "banana", "mango"}
mySet.add("orange")
mySet 
```

```python
#updating a set with multiple elements
mySet = {"apple", "banana", "mango"}
mySet.update(["orange", "kiwi"])
mySet
```

```python
#remove the element in the list
mySet = {"apple", "banana", "mango", "cherry", "kiwi", "orange"}
mySet.remove("banana")  
mySet
```

```python
#discard the element in the list
mySet = {"apple", "banana", "mango", "cherry", "kiwi", "orange"}
mySet.discard("banana")
mySet
```

```python
#pop 
mySet = {"apple", "banana", "mango", "cherry", "kiwi", "orange"}
mySet.pop()
mySet
```

```python
#clearing the element 
mySet = {"apple", "banana", "mango", "cherry", "kiwi", "orange"}
mySet.clear()
mySet
```

```python
mySet = {"apple", "banana", "mango", "cherry", "kiwi", "orange"}
mySet.discard("apple" , "banana") 
mySet
```

```python
mySet = {"apple", "banana", "mango", "cherry", "kiwi", "orange"}
mySet.remove("apple" , "banana") 
mySet
```

```python
mySet = {"apple", "banana", "mango", "cherry", "kiwi", "orange"}
mySet.pop("apple" , "banana") 
mySet
```

```python
#dosent store duplicate values
mySet = {"apple", "banana", "mango", "cherry", "kiwi", "orange", "apple"}
mySet
```

```python
#set union 
set1 = {1, 2, 3}
set2 = {3, 4, 5} 
set1.union(set2)
```

```python
#set intersection 
set1 = {1, 2, 3}
set2 = {2, 3, 4, 5}
set1.intersection(set2)
```

```python
#set difference 
set1 = {1, 2, 3}
set2 = {3, 4, 5}
set1.difference(set2)
```

```python
#symtric difference
set1 = {1, 2, 3}
set2 = {3, 4, 5}
set1.symmetric_difference(set2)
```
