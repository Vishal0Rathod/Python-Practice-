```python
# Accessing tuple elements
myTuple = ("apple", "banana", "mango")
myTuple[0]
```

```python
#slicing Tuple 
myTuple = ("apple", "banana", "mango", "cherry", "kiwi", "orange")
myTuple[2:5]
```

```python
#search the index of the element in the list
myTuple = ("apple", "banana", "mango", "cherry", "kiwi", "orange")
myTuple.index("mango")
```

```python
#count the number of times an element appears in a tuple
myTuple = ("apple", "banana", "mango", "cherry", "kiwi", "orange")
myTuple.count("mango")
```

```python
#add elements to a tuple
tuple1 = ("apple", "banana")
tuple2 = ("mango", "orange")
newTuple = tuple1 + tuple2
print(newTuple)
```
