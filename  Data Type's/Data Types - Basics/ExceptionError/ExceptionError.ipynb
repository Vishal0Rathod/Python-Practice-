```python
# ValueError:
''' Providing an inappropriate value to a function. or type conversion'''

x = int("Rajesh") #cannot convert "abc" to intiger

print(x)
```

```python
# IndexError:
''' Accessing an index that is out of range.'''

my_list = ['a','r','d']

my_list[-5] #index out of the range
```

```python
# KeyError: 
'''Accessing a key that does not exist in a dictionary.'''

my_dict = {"a": 1, "b": 2}

print(my_dict["c"]) # Key 'c' does not exist
```
