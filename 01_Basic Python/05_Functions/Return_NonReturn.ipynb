```python
# Use of yield Keyword

def newFunc(a):
    for i in range(1,a):
        yield i
values_1 =newFunc(10)

print(next(values_1))
print(next(values_1))
print(next(values_1))
```

```python
def newFunc_01(a):
    for i in range(1,a):
        return i

vlaues = newFunc_01(10)
print(vlaues)
print(vlaues)
print(vlaues)
```
