```python
'''nested loop'''
List_01 = ['Red', 'Yellow', 'Green']   
List_02 = ['White', 'Black', 'Yellow']   

for i in List_01: 
    for j in List_02: 
            if i == j: 
                print("Match Found")
            print(i,j,) 
```

```python
'''Nested Loop'''
List_01 = ['Red', 'Yellow', 'Green']
List_02 = ['White', 'Black', 'Yellow']
List_03 = ['White', 'Purple', 'Yellow']

for i in List_01:
    for j in List_02:
        for k in List_03:
            if i == j == k:
                print("Match Found:", i,j,k)
```
