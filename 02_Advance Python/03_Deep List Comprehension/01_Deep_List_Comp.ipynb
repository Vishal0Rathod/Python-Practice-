```python
NewList = []
for i in range(15):
    if i<5:
        NewList.append(i*2)
    elif i<10:
        NewList.append(i/2)
    else:
        NewList.append(i+2)
NewList
```

```python
'''Apply Filter'''
fruitlist = [1,2,3,4,5,6,7,8]
def extractonlyEvenNumber(i):
    return i % 2 == 0 

abc = filter(extractonlyEvenNumber, fruitlist)
print(abc)
print(dir(abc))

for i in abc:
    print(i)
```
