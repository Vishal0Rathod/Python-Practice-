```python
newList = []
for i in range(0,10):
    if i<2 or i>5:
        newList.append(i)
newList
```

```python
[i for i in range(0,10) if i<2 or i>5]
```

```python
NewList = []
for i in range(0,10):
    if i%2==0 and i<5:
        NewList.append(i)
NewList
```

```python
#And Conddition 
[i for i in range(0,10) if i%2==0 and i>5]
```

```python
NewList = []
for i in range(15):
    if i<5:
        NewList.append(i*2)
    elif i<10:
        NewList.append(i/2)
    else:
        NewList.append(i+2)
NewList
```

```python
[i*2 if i<5 else i/2 if i<10 else i+2 for i in range(15)]
```
