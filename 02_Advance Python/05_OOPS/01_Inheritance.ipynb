## Inheritance

```python
class A:
    def method_class_A(self):
        return "This Is Parent Class Method"

class B:
    def method_class_B(self):
        return "This Is Child Class Method"

p_A = A()
p_A.method_class_A()
```

```python
#Single Inheritance 
class ParentClass:
    def Parent_Class_Method(self):
        return "Parent Class Method I Have 50 Acres Of Land"

class ChildClass(ParentClass):
    def Child_Class_Method(self):
        return "Child Class MethodI Have Maybach"

Parent_Obj_01 = ParentClass()    
Parent_Obj_01.Parent_Class_Method()    
```

```python
Child_Obj_01 = ChildClass()
print(Child_Obj_01.Child_Class_Method())
Child_Obj_01.Parent_Class_Method()
```

# Multilevel Inheritance

```python
class GrandParentsClass:
    def GP_Class_Method(self):
        return "GrandParents Class Method I Have 100 Acres Of Land"

class ParentClass(GrandParentsClass):
    def Parent_Class_Method(self):
         return "Parent Class Method I Have 50 Acres Of Land"

class ChildClass(ParentClass):
    def Child_Class_Method(self):
        return "Child Class MethodI Have Maybach"
    
GP_Obj_01 = GrandParentsClass()
print(GP_Obj_01.GP_Class_Method())


Parent_Obj_01 = ParentClass()
print(Parent_Obj_01.Parent_Class_Method())


Child_Obj_01 = ChildClass()
print(Child_Obj_01.Child_Class_Method())
```

```python
#single Inheritance 
class Zomato:
    def deliver_food(self):
        return "Zomato - I Can Diliver Food In Minutes!"

class Blinkit(Zomato):
    def deliver_groceries(self):
        return "Blinkit - I Can Diliver groceries And Items In Minutes!"
'''    Inherting Quick Dilvery Function '''


Parent_Obj_01 = Zomato()
print(Parent_Obj_01.deliver_food())

Child_Obj_01 = Blinkit()
print(Child_Obj_01.deliver_groceries())

```

```python
#MultiLevel Inheritance 

class Tesla:
    def Grand_Parent_Company(self):
        return "Tesla - Parent company founded by Elon Musk"
    '''Elon Musk Ki Company'''


class SpaceX(Tesla):
    def Parent_Company(self):
        return "SpaceX - Builds rockets and operates the Starlink satellite network"
'''Inherting Name From Grand Parent Company'''

class StarLink(SpaceX):
    def Child_Company(self):
        return "StarLink - Provides satellite-based internet connectivity to customers"
    '''Inherting Satellite Connectivity To Provide Network Indian Customers'''


GP_Obj_01 = Tesla()
print(GP_Obj_01.Grand_Parent_Company())

Parent_Obj_01 = SpaceX()
print(Parent_Obj_01.Parent_Company())

Child_Obj_01 = StarLink()
print(Child_Obj_01.Child_Company())

```

```python
#Multiple Inheritance 

class ElectricVehical:
    def electric_mode(self):
        return "Electric - Car Runs Using Electric Power"  

class PetrolVehical:
    def petrol_mode(self):
        return "Petrol - Car Runs Using Petrol"

class HybridVehical(ElectricVehical , PetrolVehical):
    def drive(self):
        return "Hybrid - Hybrid Car Can Use Both Powers "

car_obj_01 = ElectricVehical()
print(car_obj_01.electric_mode())

car_obj_01 = PetrolVehical()
print(car_obj_01.petrol_mode())

car_obj_01 = HybridVehical()
print(car_obj_01.drive())
```
