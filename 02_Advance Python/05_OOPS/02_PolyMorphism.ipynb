# Method Overloading


```python
'''Simple Method Without Overloading'''
class myClass: 
    def sum_4(self, a,b,c,d):
        return a+b+c+d
    def sum_3(self, a,b,c):
        return a+b+c
    def sum_2(self, a,b):
        return a+b

my_obj = myClass()
print(my_obj.sum_4(1,2,3,4))
print(my_obj.sum_3(1,2,3))
print(my_obj.sum_2(1,2))
```

```python
'''With Overloading'''

class myClass:
    def sum_Overloaded(self, a=None, b=None, c=None, d=None):
        if a != None and b!= None and c!=None and d!=None:
            return a+b+c+d
        if a != None and b!= None and c!=None:
            return a+b+c
        if a != None and b!= None:
            return a+b
        else:
            return "Please pass atleast two values for sum"

my_obj = myClass()
my_obj.sum_Overloaded(1,2,3,4)
```

# Method Overriding

```python
class Animal:
    def sound(self):
        return "Animal Makes Soung"
class Dog(Animal):
    def sound(self):
        return "Barks"
class Cat(Animal):
    def sound(self):
        return "Meow"

Dog_Obj = Dog()
Dog_Obj.sound()
```

```python
class P_C_Interest:
    def interest(self):
        return 0.11

class C_C_Home_Interest(P_C_Interest):
    def interest(self):
        return 0.08

Home_oBJ = C_C_Home_Interest()
Home_oBJ.interest()
```

```python
#use of Super keyword
class P_C_Interest:
    def interest(self):
        return 0.11

class C_C_Home_Interest(P_C_Interest):
    def interest(self):
        return 0.11

    class Random_Interest(P_C_Interest):
        def interest(self):
            return super().interest()  #calls the interest method from super class
        print(0.08)

Home_oBJ = C_C_Home_Interest()
Home_oBJ.interest()
```
