# Enumerate

```python
#example of enumerate 
List_Fruit = ["Apple", "Banana", "Cherry"]
enue_fruits = enumerate(List_Fruit)

for index,value in enue_fruits:
    print(index,value)
```

```python
#example of enumerate 
List_Fruit = ["Apple", "Banana", "Cherry"]
for j, i in enumerate(List_Fruit):
    print(j,i)
```
