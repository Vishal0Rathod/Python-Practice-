```python
class Dog:
    def __init__(self, dog_Age):
        self.dog_Age = dog_Age #Instance Variable 

    def abc(self): #instance Variable
        print(self.dog_Age)

    def my_Dog_details(self, colour):
        print(f"My dog has {colour} colour")


dog_1 = Dog(14)
dog_1.abc()
dog_1.my_Dog_details("Brown")
```

```python
class Banking:
    Int_Rate = 0.08 #8% class variable

    @classmethod
    def change_int_Rate(cls):
        cls.Int_Rate = cls.Int_Rate + (cls.Int_Rate * 0.02)
        print(cls.Int_Rate)

print(Banking.Int_Rate)
Banking.change_int_Rate()
```
