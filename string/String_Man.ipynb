```python
# case change 
txt1 = "Hello My Friends" 
txt = txt1.upper() 
print(txt)


txt2 = "Hello My Friends" 
txt = txt2.lower() 
print(txt)


```

```python
#isnumeric
txt = "565543"
txt.isnumeric() 
if txt.isnumeric() == True:
  print("Yes, all characters are numeric.")
```

```python
#isdigit
txt = "50800" 
txt.isdigit() 
if txt.isdigit() == True: 
  print("Yes, all characters are digits.")

```

```python
#extracting string 
new_txt = "Rajesh40"
new_value = ""

for i in new_txt:
    if i.isalpha() == True: 
        new_value = new_value + i 

        print(new_value) 
```
